```python

import numpy as np
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

# 1. Create Dataset
data = {
    'Gender': ['Male', 'Female', 'Male', 'Female'],
    'Smoker': [1, 0, 1, 0],
    'Color': ['Red', 'Blue', 'Red', 'Green'],
    'Education': ['School', 'College', 'Graduate', 'College']
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# 2. Binary Attribute Dissimilarity
binary = df['Smoker'].values
binary_matrix = np.zeros((len(binary), len(binary)))

for i in range(len(binary)):
    for j in range(len(binary)):
        binary_matrix[i, j] = 0 if binary[i] == binary[j] else 1

print("\nBinary Dissimilarity Matrix:")
print(binary_matrix)

# 3. Nominal Attribute Dissimilarity
nominal = df['Color'].values
nominal_matrix = np.zeros((len(nominal), len(nominal)))

for i in range(len(nominal)):
    for j in range(len(nominal)):
        nominal_matrix[i, j] = 0 if nominal[i] == nominal[j] else 1

print("\nNominal Dissimilarity Matrix:")
print(nominal_matrix)

# 4. Ordinal Attribute Dissimilarity
ordinal = df['Education'].values

categories = ['School', 'College', 'Graduate']

rank = {value: i for i, value in enumerate(categories)}
normalized = np.array([rank[x] / (len(categories)-1) for x in ordinal])

ordinal_matrix = np.zeros((len(ordinal), len(ordinal)))

for i in range(len(ordinal)):
    for j in range(len(ordinal)):
        ordinal_matrix[i, j] = abs(normalized[i] - normalized[j])

print("\nOrdinal Dissimilarity Matrix:")
print(ordinal_matrix)
```
